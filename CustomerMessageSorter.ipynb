```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")

labels = ["billing problem", "technical problem",
          "wants to cancel", "general question", "refund request", "shipping problem"]

messages = [
    "I was charged twice this month, please fix it.",
    "My internet keeps disconnecting every evening.",
    "I want to end my contract at the end of the month.",
    "Do you offer a plan with streaming included?",
    "I want my money back",
    "I am not available tommorow to recive my order can u ship the order next week?",
    "When will preorders for iphone 18 pro max be available?",

]

for text in messages:
    result = classifier(text, candidate_labels=labels)
    print(text)
    print("  ->", result["labels"][0], f"({result['scores'][0]:.0%})\n")
```
